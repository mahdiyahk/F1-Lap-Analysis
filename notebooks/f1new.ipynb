```python
# reload races '\N' means missing data
races = pd.read_csv('../data/races.csv', na_values='\\N')
races.head()
```

```python
results = pd.read_csv('../data/results.csv', na_values='\\N')
drivers = pd.read_csv('../data/drivers.csv', na_values='\\N')
```

```python
# confirm /N values were correctly connverted to real nulls
races.isnull().sum()
```

```python
# remove url column - not useful for analysis
races = races.drop(columns=['url'])
drivers = drivers.drop(columns=['url'])

# drop columns - not needed for performance analysis
races = races.drop(columns=['fp1_date', 'fp1_time', 'fp2_date', 'fp2_time',
                             'fp3_date', 'fp3_time', 'quali_date', 'quali_time',
                             'sprint_date', 'sprint_time'])
races.head()

# checking if columns have been removed
races.columns
```

```python
# creates clear indicator for driver finishing race
results['finished'] = results['statusId'] == 1

# checks if finished split looks reasonable
results['finished'].value_counts()

# recheck - finish rate for most recent year
recent = results.merge(races[['raceId', 'year']], on='raceId')
recent[recent['year'] == 2024]['finished'].value_counts()
```

```python
# check what statusId values show up when positionText is 'D'
results[results['positionText'] == 'D']['statusId'].unique()

# load status lookup file, treating '\N' as missing like the others
status = pd.read_csv('../data/status.csv', na_values='\\N')
status.head()
```
